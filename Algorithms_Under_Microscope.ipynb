**Insertion Sort Algorithm  and Implementation**
*   Start with the second element as the first element is assumed to be sorted.

*   Compare the second element with the first if the second is smaller then swap them.


*   Move to the third element, compare it with the first two, and put it in its correct position
*   Repeat until the entire array is sorted.



The following implementation is using python and numpy library







```python
import numpy as np
import time

# lists = np.array([6,4,3,6,7,8,3,5,7,4])
lists = np.random.randint(0, 1_000, size=1_000)

# print(lists)



# print(lists[n-1])

start = time.perf_counter()
def sort(lists):

  z = lists.size
  for i in range(1, z):
    n = i - 1
    const = lists[i]


    while n >= 0 and const < lists[n]:
      lists[n + 1] = lists[n]
      n -= 1
    lists[n+1] = const

  # print(lists)






sort(lists)


end = time.perf_counter()

print("Runtime:", end - start, "seconds")
```

**Quick Sort Algorithm and Implementation**

1. Choose a Pivot: Select an element from the array as the pivot. The choice of
pivot can vary (e.g., first element, last element, random element, or median).

2. Partition the Array: Re arrange the array around the pivot. After partitioning, all elements smaller than the pivot will be on its left, and all elements greater than the pivot will be on its right.

3. Recursively Call: Recursively apply the same process to the two partitioned sub-arrays.
Base Case: The recursion stops when there is only one element left in the sub-array, as a single element is already sorted.:

```python

lists =[6,4,3,7,8]
#lists = np.random.randint(0, 1_000, size=1_000)

print(lists)




def quicksort(lists):

  if len(lists) <= 1:
    return lists

  small = []
  large = []

  pivot = lists[len(lists) - 1]

  for i in range(len(lists) - 1):
    if lists[i] < pivot:
      small.append(lists[i])
    else:
      large.append(lists[i])

  sorted_small = quicksort(small)
  sorted_large = quicksort(large)

  return sorted_small + [pivot] + sorted_large

quicksort(lists)


```
