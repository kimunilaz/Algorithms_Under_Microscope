**Insertion Sort Algorithm  and Implementation**
*   Start with the second element as the first element is assumed to be sorted.

*   Compare the second element with the first if the second is smaller then swap them.


*   Move to the third element, compare it with the first two, and put it in its correct position
*   Repeat until the entire array is sorted.



The following implementation is using python and numpy library

l will user timer to record the runtime of each algorithm and also generate big lists for better visualization of runtime.

l am measuring runtime in seconds.









```python

import time

#lists to accumulate average runtime for graph plotting
nlogn = []
n_squared = []
n_squared_sorted = []

#sorting function

def insertion_sort(lists):


  for i in range(1, len(lists)):
    n = i - 1
    const = lists[i]


    while n >= 0 and const < lists[n]:
      lists[n + 1] = lists[n]
      n -= 1
    lists[n+1] = const
  return lists
```

**Quick Sort Algorithm and Implementation**

1. Choose a Pivot: Select an element from the array as the pivot. The choice of
pivot can vary (e.g., first element, last element, random element, or median).

2. Partition the Array: Re arrange the array around the pivot. After partitioning, all elements smaller than the pivot will be on its left, and all elements greater than the pivot will be on its right.

3. Recursively Call: Recursively apply the same process to the two partitioned sub-arrays.
Base Case: The recursion stops when there is only one element left in the sub-array, as a single element is already sorted.:

```python


def quicksort(lists):

  if len(lists) <= 1:
    return lists

  small = []
  large = []

  pivot = lists[len(lists) -1 ]

  for i in range(len(lists) - 1):
    if lists[i] < pivot:
      small.append(lists[i])
    else:
      large.append(lists[i])

  sorted_small = quicksort(small)
  sorted_large = quicksort(large)

  return (sorted_small + [pivot] + sorted_large)


```

**Predictions**



*   Insertion sort, random input: time should roughly quadruple when n doubles (O(n²)).

*   Quicksort, random input: time should roughly double when n doubles, with a small extra factor from the log term (O(n log n)).

*   Insertion sort, already-sorted input: should be far faster than on random input, the inner loop never has to shift elements, so it behaves close to O(n).

**Discussion**

**Insertion sort, random input**

Ratios cluster around 4 (3–4.5), matching O(n²). Some noise is expected, wall-clock timing is more affected by system jitter at smaller n, where each run only takes milliseconds.



**Insertion sort, already-sorted input**

Ratios cluster near 2 (aside from one blip at n=1000→2000, likely noise since those runs took under a millisecond). A ratio near 2 means linear growth, not quadratic. This is the most interesting result: insertion sort's inner while loop shifts elements only when they're out of order. On sorted input nothing is ever out of order, so each element costs one comparison instead of up to i comparisons, the algorithm degrades from its O(n²) worst case down to O(n) on this input.


**Quicksort, random input**

Ratios settle around 2 from n=2000 onward, matching O(n log n). The first ratio (0.77, i.e. n=1000 ran faster than n=500) isn't real sub-linear behavior, both runs took under 1.3 ms, where interpreter and system overhead dominate the actual sorting work.

**Conclusion**

The theory holds: insertion sort roughly quadruples with n (O(n²)), quicksort roughly doubles (O(n log n)), and insertion sort on sorted input drops to linear growth because its inner loop does no work when nothing needs shifting, the one case where its O(n²) label doesn't describe what actually happens.
In practice, quicksort is the better default: the O(n log n) vs O(n²) gap is already about 170× at n=8000 and only grows from there. Insertion sort is worth choosing only for small n or data that's already nearly sorted.
