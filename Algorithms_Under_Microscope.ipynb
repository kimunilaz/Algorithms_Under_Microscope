```python
import numpy as np
import time

# lists = np.array([6,4,3,6,7,8,3,5,7,4])
lists = np.random.randint(0, 1_000, size=1_000)

# print(lists)



# print(lists[n-1])

start = time.perf_counter()
def sort(lists):


  for i in range(1, lists.size):
    n = i - 1
    const = lists[i]


    while n >= 0 and const < lists[n]:
      lists[n + 1] = lists[n]
      n -= 1
    lists[n+1] = const

  # print(lists)






sort(lists)


end = time.perf_counter()

print("Runtime:", end - start, "seconds")
```
