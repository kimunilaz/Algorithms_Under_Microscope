```python
import numpy as np

lists = np.array([6,4])

print(lists)

n = 1

# print(lists[n-1])
while n < lists.size:
  if lists[n-1] > lists[n]:
    lists[n], lists[n-1] = lists[n-1], lists[n]
    n = n + 1
  else:
    n = n + 1


print(lists)

```
